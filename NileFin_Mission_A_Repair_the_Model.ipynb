## 2. Compare three representations

Both lists and dictionaries can have a variable name, such as `customer`. Lists access values by position; dictionaries use keys to label values. Dataclasses provide named fields.

### A. List

Example:

`customer = ["C101", "Amina", "Kampala"]`

**Advantages**

- Simple.
- Short.
- Easy to use for very small data.

**Disadvantages**

- We must remember what positions 0, 1 and 2 mean.
- It is easy to put values in the wrong position.

### B. Dictionary

Example:

`customer = {"id": "C101", "name": "Amina", "city": "Kampala"}`

**Advantages**

- Descriptive keys make the data easier to understand.
- We can use `customer["name"]` instead of `customer[1]`.

**Disadvantages**

- Keys can be mistyped.
- Takes slightly more code than a list.

### C. Dataclass

Example:

`Customer("C101", "Amina", "Kampala")`

**Advantages**

- Gives customer records a clear structure.
- Fields have meaningful names.
- Easier to maintain as the system grows.

**Disadvantages**

- Requires defining a class first.
- May be unnecessary for a small, one-off script.


## 3. Implement two representations

We will implement:

1. **List**
2. **Dataclass**


### Representation 1 ? List


```python
customers_list = [
    ["C101", "Amina", "Kampala"],
    ["C102", "Joan", "Entebbe"],
    ["C103", "Amina", "Jinja"]
]

# Print each customer's name
for customer in customers_list:
    print(customer[1])
```

### Representation 2 ? Dataclass


```python
from dataclasses import dataclass

@dataclass
class Customer:
    customer_id: str
    name: str
    city: str

customers_dataclass = [
    Customer("C101", "Amina", "Kampala"),
    Customer("C102", "Joan", "Entebbe"),
    Customer("C103", "Amina", "Jinja")
]

# Print each customer's name
for customer in customers_dataclass:
    print(customer.name)
```

## 4. When the list works well

A **list** works well for a small, quick script with a few fixed values.

For example, printing all records is simple:


```python
for customer in customers_list:
    print(customer)
```

A list needs little setup. This may be enough for a temporary script.


## 5. Where the list can cause mistakes

In a list, each value's meaning depends on its **position**.

The intended order is:

`[customer_id, name, city]`

Someone could accidentally swap the ID and name:


```python
# This is wrong, but Python accepts it.
wrong_customer = ["Amina", "C105", "Kampala"]

print("Customer ID:", wrong_customer[0])
print("Name:", wrong_customer[1])
print("City:", wrong_customer[2])
```

Python does not know that `"Amina"` should be the name, not the customer ID.

Using named fields in a dataclass makes the meaning clearer:


```python
correct_customer = Customer(
    customer_id="C105",
    name="Amina",
    city="Kampala"
)

print("Customer ID:", correct_customer.customer_id)
print("Name:", correct_customer.name)
print("City:", correct_customer.city)
```

## 6. What I would ship today

I would use a **dataclass** for the customer model.

NileFin already has customer records, transactions, support records, duplicates and missing values. Management is also asking about prediction, so the data structure is likely to grow.

Instead of remembering:

`customer[0]`, `customer[1]`, `customer[2]`

we can write:

`customer.customer_id`, `customer.name`, `customer.city`

This makes the code easier to read and helps reduce mistakes.


# Runnable proof

This example uses the dataclass to find a customer by ID, find customers by city, and add a customer.


```python
from dataclasses import dataclass

@dataclass
class Customer:
    customer_id: str
    name: str
    city: str

customers = [
    Customer("C101", "Amina", "Kampala"),
    Customer("C102", "Joan", "Entebbe"),
    Customer("C103", "Amina", "Jinja")
]

# Find customer by ID
search_id = "C102"

for customer in customers:
    if customer.customer_id == search_id:
        print("Found:", customer)

# Find customers in a city
search_city = "Jinja"

for customer in customers:
    if customer.city == search_city:
        print("Customer in Jinja:", customer)

# Add a customer
customers.append(Customer("C104", "Peter", "Kampala"))

print("\nAll customers:")
for customer in customers:
    print(customer)
```

# Trade-off

A dataclass needs more setup than a list.

A list may be quicker for a small, temporary script. A dataclass offers a clearer structure for a system that will grow and need maintenance.
